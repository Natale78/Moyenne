```python
#version5
from tkinter import *
from tkinter import ttk

fenetre = Tk()
fenetre.title("calcul moyenne")
fenetre.geometry("880x600")  # Hauteur réduite pour voir le scroll en action

# 1. Création d'un conteneur avec un Canvas et une Scrollbar
main_frame = Frame(fenetre)
main_frame.pack(fill=BOTH, expand=1)

my_canvas = Canvas(main_frame)
my_canvas.pack(side=LEFT, fill=BOTH, expand=1)

# Ajout de la barre de défilement verticale
my_scrollbar = ttk.Scrollbar(main_frame, orient=VERTICAL, command=my_canvas.yview)
my_scrollbar.pack(side=RIGHT, fill=Y)

# Configuration du Canvas
my_canvas.configure(yscrollcommand=my_scrollbar.set)
my_canvas.bind('<Configure>', lambda e: my_canvas.configure(scrollregion=my_canvas.bbox("all")))

# 2. Création d'un sous-cadre (frame) à l'intérieur du Canvas qui contiendra les widgets
content_frame = Frame(my_canvas)
my_canvas.create_window((0, 0), window=content_frame, anchor="nw")

# Listes et variables
ressources = [f"R{i}" for i in range(101, 116)]
sae = [f"SAE{i}" for i in range(11, 17)]
entrees = []

def notes(c, liste):
    for index, val in enumerate(liste):
        etiquette = ttk.Label(content_frame, text=val)
        etiquette.grid(column=c, row=index, padx=10, pady=10)
        saisie = ttk.Entry(content_frame)
        saisie.grid(column=c+1, row=index, padx=10, pady=10)
        entrees.append(saisie)

notes(0, ressources)
notes(2, sae)

def recup():
    saisies = [i.get() for i in entrees]
    return saisies

bouton = ttk.Button(content_frame, text="Calculer", command=recup)
bouton.grid(column=2, row=16, padx=10, pady=20)

fenetre.mainloop()
```
