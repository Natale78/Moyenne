# Calcul des moyennes d'UE - BUT 1 R&T (semestre 1)

**Moyenne d'une UE** = somme(note × coef) / somme(coefs)

**Règles de passage**
- Les 3 UE ≥ 10 : année validée, passage en BUT 2.
- 2 UE ≥ 10 et la 3e entre 8 (inclus) et 10 (exclu) : année non validée, mais passage en BUT 2 autorisé (UE à valider en BUT 2).
- 2 UE ≥ 10 et la 3e < 8 : année non validée, passage non autorisé, redoublement possible selon l'avis du jury.
- 2 ou 3 UE < 10 : année non validée, passage en BUT 2 exclu.

## Version 1 : saisie au clavier

```python
# Version 1 : notes et coefficients saisis au clavier (pour UNE UE)

def calcul_moyUE():
    nb = int(input("Combien de ressources / SAÉ comptent dans cette UE ? "))
    somme = 0
    total_coefs = 0
    for i in range(1, nb + 1):
        coef = int(input(f"Coefficient de la matière n°{i} : "))
        note = float(input(f"Note de la matière n°{i} : "))
        somme = somme + coef * note
        total_coefs = total_coefs + coef
    return somme / total_coefs

moyenne = calcul_moyUE()
print(f"Moyenne de l'UE : {moyenne:.2f}")
```

## Version 2 : listes

```python
# Version 2 : notes et coefficients rangés dans des listes (semestre 1)

# Une liste avec toutes les matières, et une liste de coefficients par UE
# (même ordre : le coefficient n°i correspond à la matière n°i, 0 = ne compte pas dans l'UE)
matieres = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]
coef1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
coef2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
coef3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

# Saisie des notes
notes = []
for matiere in matieres:
    notes.append(float(input(f"Note de {matiere} : ")))

def moyenne_UE(notes, coefs):
    somme = 0
    total_coefs = 0
    for note, coef in zip(notes, coefs):
        somme = somme + note * coef
        total_coefs = total_coefs + coef
    return somme / total_coefs

print(f"UE1 = {moyenne_UE(notes, coef1):.2f}")
print(f"UE2 = {moyenne_UE(notes, coef2):.2f}")
print(f"UE3 = {moyenne_UE(notes, coef3):.2f}")
```

```python
# Version 4 : graphiques + décision du jury (semestre 1)
import matplotlib.pyplot as plt

# Pour chaque UE : {matière: coefficient}
COEFS = {
    "UE1": {"R101": 10, "R102": 10, "R103": 7, "R104": 7, "R106": 5, "R108": 6, "R110": 5, "R111": 4, "R112": 2, "R113": 5, "R114": 5, "SAE11": 20, "SAE12": 20, "SAE16": 7},
    "UE2": {"R101": 4, "R103": 2, "R104": 8, "R105": 6, "R110": 5, "R111": 5, "R112": 2, "R113": 9, "R114": 9, "R115": 3, "SAE13": 29, "SAE16": 7},
    "UE3": {"R101": 4, "R103": 2, "R106": 5, "R107": 15, "R108": 6, "R109": 4, "R110": 5, "R111": 5, "R112": 2, "R115": 3, "SAE14": 20, "SAE15": 20, "SAE16": 7},
}

# Toutes les matières du semestre 1
matieres = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

def saisir_note(matiere):
    note = float(input(f"Note de {matiere} : "))
    while note < 0 or note > 20:
        note = float(input("La note doit être entre 0 et 20, recommence : "))
    return note

def moyenne_UE(notes, coefs_ue):
    somme = 0
    total_coefs = 0
    for matiere, coef in coefs_ue.items():
        somme = somme + notes[matiere] * coef
        total_coefs = total_coefs + coef
    return somme / total_coefs

def decision_jury(moyennes):
    """moyennes : liste des 3 moyennes d'UE. Renvoie la décision sous forme de texte."""
    nb_ok = 0                      # nombre d'UE >= 10
    for moy in moyennes:
        if moy >= 10:
            nb_ok = nb_ok + 1

    if nb_ok == 3:
        return "Année validée : passage en BUT 2 autorisé."
    elif nb_ok == 2:
        if min(moyennes) >= 8:     # la 3e UE est entre 8 (inclus) et 10 (exclu)
            return "Année non validée mais passage en BUT 2 autorisé : il faudra valider l'UE manquante en BUT 2."
        else:                      # la 3e UE est < 8
            return "Année non validée, passage en BUT 2 non autorisé : redoublement possible selon l'avis du jury."
    else:
        return "Année non validée : passage en BUT 2 exclu."

def couleur(note):
    if note < 8:
        return "red"
    elif note < 10:
        return "orange"
    else:
        return "green"

# --- Saisie des notes : {matière: note} ---
notes = {}
for matiere in matieres:
    notes[matiere] = saisir_note(matiere)

# --- Moyennes des 3 UE et décision ---
noms_ue = []
moyennes = []
for ue, coefs_ue in COEFS.items():
    moy = moyenne_UE(notes, coefs_ue)
    noms_ue.append(ue)
    moyennes.append(moy)
    print(f"{ue} : {moy:.2f}")

print(decision_jury(moyennes))

# --- Graphique 1 : moyennes des UE ---
plt.figure()
plt.bar(noms_ue, moyennes, color=[couleur(m) for m in moyennes])
plt.axhline(10, color="black", linestyle="--")      # barre des 10
plt.ylim(0, 20)
plt.ylabel("Moyenne")
plt.title("Moyennes des UE")
plt.show()

# --- Graphique 2 : détail d'une UE ---
numero = input("Quelle UE veux-tu détailler (1, 2 ou 3) ? ")
while numero not in ("1", "2", "3"):
    numero = input("Tape 1, 2 ou 3 : ")
ue = "UE" + numero

noms = []
valeurs = []
for matiere in COEFS[ue]:          # seulement les matières qui comptent dans l'UE
    noms.append(matiere)
    valeurs.append(notes[matiere])

plt.figure()
plt.bar(noms, valeurs, color=[couleur(v) for v in valeurs])
plt.axhline(10, color="black", linestyle="--")
plt.ylim(0, 20)
plt.xticks(rotation=90)
plt.ylabel("Note")
plt.title(f"Détail de {ue}")
plt.show()
```
